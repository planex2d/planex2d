# How AI is used in this project

Yes, planex2d is developed with AI agents. They do not make the decisions, though: they implement what has been decided,
and humans review what they write.

## Development

The API, the skeleton of the code and the algorithms are decided and sketched up front by humans. An agent then
implements one decided piece at a time. The coding style has strict rules, and every line that is implemented is
reviewed by a human: it has to make sense and be clean, readable and simple C++. The review also checks that the code
is not overly complex or overly general. planex2d uses what is known about the geometry by construction, and the code
has to show it.

## Documentation

The examples, the sections and the topics of these docs are set by humans. The code and the prose are generated by an
agent, then checked and rewritten by humans. The pictures are set by humans too: what is shown, the viewport, the angle,
the colors, etc.
